# Mathematical Operations

## Scalar Operations

```python
x = torch.tensor([1, 2, 3, 4, 5], dtype=torch.int64)
x
```

```python
# Addition
x + 2
```

```python
# Subtraction
x - 2
```

```python
# Multiplication
x * 2
```

```python
# Division
x / 3
```

```python
# Integer Division
x // 3
```

```python
# Mod
((x * 100)//3)%2
```

```python
# power
x ** 2
```

## Element wise Operation

```python
a = torch.rand(2,3)
b = torch.rand(2,3)

print(a)
print(b)
```

```python
# add
a + b
```

```python
# sub
a - b
```

```python
# multiply
a * b
```

```python
# division
a / b
```

```python
# power
a ** b
```

```python
# mod
a % b
```

```python
c = torch.tensor([1, -2, 3, -4])
```

```python
# abs
torch.abs(c)
```

```python
# negative
torch.neg(c)
```

```python
d = torch.tensor([1.9, 2.3, 3.7, 4.4])
```

```python
# round
torch.round(d)
```

```python
# ceil
torch.ceil(d)
```

```python
# floor
torch.floor(d)
```

```python
# clamp
torch.clamp(d, min=2, max=3)
```

## Reduction Operation

```python
e = torch.randint(size=(2,3), low=0, high=10, dtype=torch.float32)
e
```

```python
# sum
torch.sum(e)
```

```python
# sum along columns
torch.sum(e, dim=0)
```

```python
# sum along rows
torch.sum(e, dim=1)
```

```python
# mean
torch.mean(e)
```

```python
# mean along col
torch.mean(e, dim=0)
```

```python
# median
torch.median(e)
```

```python
# max and min
torch.max(e)
torch.min(e)
```

```python
# product
torch.prod(e)
```

```python
# standard deviation
torch.std(e)
```

```python
# variance
torch.var(e)
```

```python
# argmax
torch.argmax(e)
```

```python
# argmin
torch.argmin(e)
```
