# Creating a tensor

```python
# Using Empty
a = torch.empty(2, 3)
a
```

```python
# Check type
print(type(a))
```

```python
# Using Zeros
torch.zeros(2, 4)
```

```python
# Using ones
torch.ones(2, 4)
```

```python
# Using Rand
torch.manual_seed(42)
torch.rand(2, 4)
```

```python
# Using tensor
torch.tensor([[1, 2, 3], [4, 5, 6]])
```

```python
# arange
torch.arange(0, 20, 3)
```

```python
# linspace
torch.linspace(0, 10, 5)
```

```python
# eye
torch.eye(5)
```

```python
# full
torch.full((4, 2), 10)
```

```python
x.shape
```

```python
torch.empty_like(x)
```

# Mathematical Operations

## Scalar Operations

```python
x = torch.tensor([1, 2, 3, 4, 5], dtype=torch.int64)
x
```

```python
# Addition
x + 2
```

```python
# Subtraction
x - 2
```

```python
# Multiplication
x * 2
```

```python
# Division
x / 3
```

```python
# Integer Division
x // 3
```

```python
# Mod
((x * 100)//3)%2
```

```python
# power
x ** 2
```
