```python
import torch
```

```python
x = torch.tensor(2.0)       # input value: 2.0
target = torch.tensor(10)   # correct answer

# requires_grad = True on w means that I want the gradient for this one
# I only use requires_grad = True here, and not anywhere else because w is the only thing we are allowed to change
w = torch.tensor(3.0, requires_grad=True)       # knob, value = 3.0
```

```python
x
```

```python
target
```

```python
w
```

```python
prediction = w * x
error = prediction - target
loss = error ** 2
```

```python
loss.backward()     # walk the receipt backward
```

```python
# It means that increase w and the loss falls by about 16 for every 1 w you add
print(w.grad)
```
