```python
import torch
```

# A simple example

```python
# Create a tensor
# requires_grad = True --> It enables gradient tracking
x = torch.tensor(3.0, requires_grad=True)

# Perform the mathematical operation
y = x ** 2

# This tells Pytorch: "Work backward from y and calculate how changing x would affect y". Pytorch calculates this derivative automatically instead of me doing it by hand
y.backward()

# Retrieve the gradient
print(x.grad)
```
