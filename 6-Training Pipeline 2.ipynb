```python
import torch
```

# Step 1 - Prepare the dataset

```python
X = torch.tensor([
    [1000.0],
    [1500.0],
    [2000.0],
    [2500.0]
])
print(X)
```

```python
y = torch.tensor([
    [100000.0],
    [150000.0],
    [200000.0],
    [250000.0]
])
print(y)
```

# Step 2 - DataLoader

```python
from torch.utils.data import TensorDataset, DataLoader

dataset = TensorDataset(X, y)

# dataset --> The training data
# batch_size --> Feed two samples at once
# shuffle = True --> Randomize the order of samples
dataloader = DataLoader(dataset, batch_size=2, shuffle=True)
```

```python
dataloader
```

# Step 3 - Define the Neural Network

```python
import torch.nn as nn
```

```python
# Creates a fully connected layer with 1 input feature, 1 output feature, 1 weight, 1 bias.
# The mathematical equation is y = w*x + b. Pytorch initializes the weight and weight and bias automatically

# Syntax --> nn.Linear(in_features, out_features, bias)
# in_features --> size of each input sample
# out_features --> size of each output sample
model = nn.Linear(1, 1)
```

# Step 4 - Define a loss function

```python
# We will use Mean Squared Error
loss_fn = nn.MSELoss()
loss_fn
```

# Step 5 - Define an optimizer

```python
optimizer = torch.optim.SGD(model.parameters(), lr=0.01)

# model.parameters() --> It gives the trainable weights and biases
# SGD is the Stochastic Gradient Descent
# lr is the learning rate
```

# Step 6 - Zero the gradients

```python
optimizer.zero_grad()
```
