```python
# torch gives us tensors, mathematical operations, gradient calculations, and GPU support
import torch

# torch.nn provides tools for building neural networks.
import torch.nn as nn
```

```python
# Training data: y = 2x

# X contains the input, while y contains the correct labels

# We use torch.tensor instead of normal python list because tensor is Pytorch's fundamental data structure. It supports efficient numerical calculations, can run on compatible GPU's, and can participate in automatic gradient calculations
X = torch.tensor([[1.0], [2.0], [3.0], [4.0]])
y = torch.tensor([[2.0], [4.0], [6.0], [8.0]])
```

```python
print(X.shape)      # [4, 1] --> It means we have 4 examples and 1 means that each example has one input feature
print(X)
```

```python
y
```

## Create the neural network

```python
# nn.Linear(in_features, out_features) creates a linear layer. In our case there is one input feature and one output feature
# The mathematical equation is: y = wx + b
# The model initially starts with weight and bias values that Pytorch initializes for us. They are not necessarily the correct values.
# Our goal is to learn y = 2x
model = nn.Linear(1, 1)
```

```python
# Inspect the initial parameters:
print(model.weight)
print(model.bias)
```

## Choose a loss function and an optimizer

```python
# Define Loss function. In this case, its mean squared error
loss_function = nn.MSELoss()

# Choose an optimizer
# torch.optim.SGD is the optimization algorithm that adjusts the model's parameters to reduce the loss
# model.parameters() --> THis gives the optimizer access to the learnable parameters of our model. In this case, its weight and bias
# learning rate = 0.01
optimizer = torch.optim.SGD(model.parameters(), lr=0.01)
```

## Start the Training Loop

```python
# We are repeating the process 1000 times
for epoch in range(1000):
    # Make predictions: Pytorch passes all four inputs through the model. For each input, the model calculates: y = wx + b
    predictions = model(X)

    # Calculate the loss. This compares the model's predictions with the correct outputs.
    loss = loss_function(predictions, y)

    # Clear the old gradients: By default, PyTorch accumulates gradients. When we calculate gradients repeatedly, it adds them to the existing gradient values instead of automatically replacing them. We clear the old gradients before calculating the new ones, ensuring that the optimizer uses the gradients from the current training step rather than an unintended accumulation from previous steps.
    optimizer.zero_grad()

    # Calculate gradients through backpropagation. This is where PyTorch's automatic differentiation system, called Autograd, does its work.
    loss.backward()

    # Update the parameters -> Now the optimizer uses the gradients to update the model's parameters. For standard SGD, the update rule is w_new = w_old - (lr * gradient)
    optimizer.step()
```

## Make Prediction after training

```python
# Test the trained model.
# torch.no_grad() --> We are only making a prediction here, not training the model. Disabling gradient tracking avoids unnecessary computation and memory usage
with torch.no_grad():
    # We provide a new input value of 5.0. Its shape is [1, 1], meaning one example with one feature. The model applies its learned weight and bias:
    print(model(torch.tensor([[5.0]])))
```
